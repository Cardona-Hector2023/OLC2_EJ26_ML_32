```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Cargar el dataset y limpiar las fechas
df_resenas = pd.read_csv('resenas_clientes_dev.csv')

# Convertimos la fecha a formato correcto de Pandas para poder agrupar por año
df_resenas['fecha_reseña'] = pd.to_datetime(df_resenas['fecha_reseña'], errors='coerce')
df_resenas['año'] = df_resenas['fecha_reseña'].dt.year

# Calculamos el largo del texto para tener una métrica numérica extra
df_resenas['largo_comentario'] = df_resenas['texto_reseña'].str.len()

# 2. CONFIGURACIÓN DEL LIENZO (1 fila, 3 gráficos clave)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --- GRÁFICO 1: Categorías de Servicio (Tus barras solicitadas) ---
conteo_cat = df_resenas['categoria_servicio'].value_counts()
conteo_cat.plot(kind='barh', ax=axes[0], color="purple", edgecolor="white")
axes[0].invert_yaxis()
axes[0].set_title("Reseñas por Categoría de Servicio", fontweight="bold")
axes[0].set_xlabel("Cantidad de Reseñas")

# --- GRÁFICO 2: Evolución en el Tiempo ---
conteo_tiempo = df_resenas['año'].value_counts().sort_index()
conteo_tiempo.plot(kind='line', ax=axes[1], marker='o', color="darkorange", linewidth=2)
axes[1].set_title("Evolución de Reseñas por Año", fontweight="bold")
axes[1].set_ylabel("Cantidad de Reseñas")
axes[1].grid(True, linestyle='--', alpha=0.6)

# --- GRÁFICO 3: Distribución del largo de los textos ---
axes[2].hist(df_resenas['largo_comentario'].dropna(), bins=20, color="teal", edgecolor="white")
axes[2].set_title("Distribución del Largo del Comentario (Caracteres)", fontweight="bold")
axes[2].set_ylabel("Frecuencia")
axes[2].set_xlabel("Número de caracteres")

plt.suptitle("Análisis Exploratorio de Datos (EDA) Real - Reseñas de Clientes", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```
