```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import pandas as pd

columnas = [
    'proyectos_completados', 'ingresos_totales', 'tarifa_hora_promedio',
    'anios_experiencia', 'tiempo_respuesta_horas', 'tasa_finalizacion',
    'calificacion_promedio', 'clientes_recurrentes', 'horas_trabajadas_mes'
]
# Esto convierte textos extraños a NaN y permite que .mean() y .median() funcionen sin romper el código
for col in columnas:
    df1[col] = pd.to_numeric(df1[col], errors='coerce')

fig, axes = plt.subplots(3, 3, figsize=(16, 12))
fig.suptitle('Distribución de Variables — Freelancers', fontsize=16, fontweight='bold', y=1.02)
axes = axes.flatten()

# Reemplaza la línea 'colores = [...]' en tu código con esto:
colores = ['#FF6B6B', '#FFD93D', '#6BCB77', '#4D96FF', '#F473B9', 
           '#FF9F43', '#00C9A7', '#845EC2', '#D65DB1']
for i, col in enumerate(columnas):
    ax = axes[i]
    
    # 2. OPTIMIZACIÓN: Se usa 'skipna=True' por si quedaron valores vacíos tras la conversión
    media   = df1[col].mean(skipna=True)
    mediana = df1[col].median(skipna=True)

    # Graficar el histograma (eliminamos valores nulos temporalmente solo para el gráfico con .dropna())
    ax.hist(df1[col].dropna(), bins=30, color=colores[i], edgecolor='white', alpha=0.85)
    
    # Líneas de media y mediana
    ax.axvline(media,   color='red',   linestyle='--', linewidth=1.5, label=f'Media: {media:.1f}')
    ax.axvline(mediana, color='black', linestyle='-',  linewidth=1.5, label=f'Mediana: {mediana:.1f}')

    # Formato estético
    ax.set_title(col.replace('_', ' ').title(), fontsize=11, fontweight='bold')
    ax.set_xlabel('Valor')
    ax.set_ylabel('Frecuencia')
    ax.legend(fontsize=8)
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('eda_distribucion_freelancers.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Aseguramos que las columnas críticas sean numéricas (Limpieza obligatoria)
columnas_a_numericas = ['proyectos_completados', 'tarifa_hora_promedio', 'calificacion_promedio']
for col in columnas_a_numericas:
    if df1[col].dtype == 'object':
        df1[col] = df1[col].astype(str).str.replace('$', '', regex=False).str.strip()
    df1[col] = pd.to_numeric(df1[col], errors='coerce')

# 2. Filtramos solo las columnas numéricas
columnas_numericas = df1.select_dtypes(include=['int64', 'float64']).columns.tolist()

# 3. ¡EL TRUCO! Normalizamos los datos (Escala de 0 a 1) para que convivan en la misma gráfica
# Esto hace que el mínimo de cada variable sea 0 y el máximo sea 1.
df_normalizado = (df1[columnas_numericas] - df1[columnas_numericas].min()) / (df1[columnas_numericas].max() - df1[columnas_numericas].min())

# 4. Creamos una única gráfica
fig, ax = plt.subplots(figsize=(12, 6))

# Dibujamos todos los boxplots juntos en 'ax'
df_normalizado.boxplot(ax=ax, 
                       rot=30, 
                       patch_artist=True,
                       boxprops=dict(facecolor="lightblue", color="steelblue"),
                       flierprops=dict(marker='o', markerfacecolor='red', markersize=5, linestyle='none'))

ax.set_title("Boxplots por Variable (Escala Normalizada 0-1) — Detección de Outliers", fontsize=14, fontweight="bold")
ax.set_ylabel("Escala Relativa")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Aseguramos (por si acaso) que las columnas numéricas estén bien convertidas en df1
columnas_a_numericas = ['proyectos_completados', 'tarifa_hora_promedio', 'calificacion_promedio']
for col in columnas_a_numericas:
    if df1[col].dtype == 'object':
        df1[col] = df1[col].astype(str).str.replace('$', '', regex=False).str.strip()
    df1[col] = pd.to_numeric(df1[col], errors='coerce')

# 2. Obtenemos la lista exacta de columnas numéricas (excluyendo IDs y categorías de texto)
columnas_numericas = df1.select_dtypes(include=['int64', 'float64']).columns.tolist()

# 3. Calculamos la matriz de correlación de Pearson
corr = df1[columnas_numericas].corr()

# 4. Diseñamos el mapa de calor (Heatmap)
fig, ax = plt.subplots(figsize=(10, 8)) # Un poco más grande para que se lean bien los nombres
im = ax.imshow(corr.values, cmap="coolwarm", vmin=-1, vmax=1)

# Añadimos la barra de color al lado
plt.colorbar(im, ax=ax, label="Correlación de Pearson")

# Configuramos las etiquetas de los ejes
ax.set_xticks(range(len(columnas_numericas)))
ax.set_yticks(range(len(columnas_numericas)))
ax.set_xticklabels(columnas_numericas, rotation=45, ha="right", fontsize=10)
ax.set_yticklabels(columnas_numericas, fontsize=10)

# 5. Dibujamos los números dentro de cada cuadrito
for i in range(len(columnas_numericas)):
    for j in range(len(columnas_numericas)):
        # Obtenemos el valor de la correlación
        valor = corr.iloc[i, j]
        # Si el valor es NaN (por datos vacíos), ponemos un guión, si no, el número
        texto = f"{valor:.2f}" if not pd.isna(valor) else "-"
        
        # Elegimos color de letra (negra o blanca) según el fondo para que se lea bien
        color_texto = "white" if abs(valor) > 0.5 else "black"
        
        ax.text(j, i, texto, ha="center", va="center", fontsize=9, color=color_texto)

ax.set_title("Matriz de Correlación entre Variables Numéricas", fontsize=14, fontweight="bold", pad=20)
plt.tight_layout()
plt.show()
```
