```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar tus datos
#df = pd.read_csv('creditguard_dataset.csv')

# 2. Definir las variables que queremos graficar (9 en total)
columnas = [
    'ingresos_mensuales', 'deuda_activa', 'monto_solicitado',
    'historial_pagos', 'utilizacion_credito', 'antiguedad_laboral',
    'atrasos_previos', 'dependientes_economicos', 'creditos_activos'
]

# 3. Crear la estructura de la cuadrícula (3 filas x 3 columnas)
# figsize=(15, 12) le da un buen tamaño para que no se amontone todo
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 12))

# 4. Aplanar la matriz de "axes" a una sola dimensión para recorrerla fácilmente
axes = axes.flatten()

# 5. Iterar sobre cada columna y graficarla en su respectivo recuadro
for i, col in enumerate(columnas):
    # Elegir colores diferentes según el tipo de variable (opcional, solo para diseño)
    if col in ['ingresos_mensuales', 'deuda_activa', 'monto_solicitado']:
        color_grafica = 'green'      # Monetarias
    elif col in ['historial_pagos', 'utilizacion_credito']:
        color_grafica = 'orange'     # Porcentajes
    elif col in ['atrasos_previos', 'dependientes_economicos', 'creditos_activos']:
        color_grafica = 'purple'     # Conteo
    else:
        color_grafica = 'teal'       # Antigüedad
        
    # Graficar en el "ax" correspondiente
    sns.histplot(df[col], kde=True, color=color_grafica, ax=axes[i])
    axes[i].set_title(f'Distribución de {col}', fontsize=10)
    axes[i].set_xlabel('')  # Quitamos el label del eje X para que se vea más limpio

# 6. Ajustar automáticamente los espacios entre gráficas para que no se encimen los textos
plt.tight_layout()

# 7. Título principal de toda la ventana
plt.suptitle('Análisis de Distribuciones - CreditGuard Dataset', y=1.02, fontsize=16, fontweight='bold')

# 8. Mostrar el resultado final
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Definir las variables para los boxplots (las 9 numéricas)
columnas = [
    'ingresos_mensuales', 'deuda_activa', 'monto_solicitado',
    'historial_pagos', 'utilizacion_credito', 'antiguedad_laboral',
    'atrasos_previos', 'dependientes_economicos', 'creditos_activos'
]

# 3. Crear la estructura de la cuadrícula (3 filas x 3 columnas)
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(16, 10))

# 4. Aplanar la matriz de posiciones para iterar fácilmente
axes = axes.flatten()

# 5. Generar un boxplot para cada variable en su respectivo recuadro
for i, col in enumerate(columnas):
    # Mantenemos el color naranja que elegiste, o puedes personalizarlo
    sns.boxplot(x=df[col], color='orange', ax=axes[i])
    
    # Configuraciones individuales por sub-gráfica
    axes[i].set_title(f'Outliers en {col}', fontsize=11, fontweight='bold')
    axes[i].set_xlabel('')  # Limpiamos el texto inferior para que no sature

# 6. Ajustar espacios entre sub-gráficas
plt.tight_layout()

# 7. Título principal de toda la figura
plt.suptitle('Detección Global de Outliers - CreditGuard Dataset', y=1.02, fontsize=16, fontweight='bold')

# 8. Mostrar el lienzo con todas las gráficas juntas
plt.show()
```
